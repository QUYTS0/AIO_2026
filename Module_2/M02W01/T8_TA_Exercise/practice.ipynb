```python
import numpy as np
import scipy.spatial.distance as distance

def compute_L2_norm(vector):

    return np.linalg.norm(vector)

def compute_dot_product(vector1, vector2):
    return vector1@vector2

def compute_cosine_similarity(vector1, vector2):
    dot_product = compute_dot_product(vector1, vector2)
    norm_v1 = compute_L2_norm(vector1)
    norm_v2 = compute_L2_norm(vector2)

    return dot_product / (norm_v1 * norm_v2)

def compute_cosine_scipy(vector1, vector2):
    return distance.cosine(vector1, vector2)


```

```python
vector = np.array([4, 5])

L2_norm = compute_L2_norm(vector)
print("The L2 norm of the vector is:", L2_norm)

v1 = np.array([0, 1, -1, 2])
v2 = np.array([2, 5, 1, 0])

dot_product = compute_dot_product(v1, v2)
print("The dot product of the two vectors is:", dot_product)
print(round(dot_product, 2))

cosine_similarity = compute_cosine_similarity(v1, v2)
print("The cosine similarity of the two vectors is: ", cosine_similarity)
print(round(cosine_similarity, 3))

cosine_similarity_scipy = compute_cosine_scipy(v1, v2)
print("The cosine similarity of the two vectors using scipy is: ", cosine_similarity_scipy)
print(round(1-cosine_similarity_scipy, 3))

```
